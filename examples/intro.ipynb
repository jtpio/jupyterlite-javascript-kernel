## Functions

```javascript
function factorial(n) {
    if (n <= 1) return 1;
    return n * factorial(n - 1);
}
```

```javascript
factorial(10)
```

```javascript
// Arrow functions
const square = x => x * x;
const numbers = [1, 2, 3, 4, 5];
numbers.map(square)
```

## Console Output

`console.log()` and `console.error()` output is captured:

```javascript
console.log("This is a log message");
console.log("Multiple", "arguments", "work", "too");
```

```javascript
console.error("This is an error message");
console.warn("This is a warning");
```

Values are printed like in Node.js, and the format specifiers like `%s`, `%d` and `%o` are replaced:

```javascript
console.log("%s was born in %d", "Ada Lovelace", 1815);
console.log({ name: "Ada", languages: ["Analytical Engine"], born: new Date(1815, 11, 10) });
```

`console.table()` shows arrays and objects as tables:

```javascript
console.table([
    { language: "JavaScript", year: 1995 },
    { language: "Python", year: 1991 },
    { language: "Julia", year: 2012 }
]);
```

The groups, counters and timers work like in the browser:

```javascript
console.group("Build");
console.log("Compiling...");
console.count("step");
console.count("step");
console.groupEnd();

console.time("loop");
let total = 0;
for (let i = 0; i < 1_000_000; i++) {
    total += i;
}
console.timeEnd("loop");
```

## Async/Await

Top-level await is supported:

```javascript
const delay = ms => new Promise(resolve => setTimeout(resolve, ms));

console.log("Starting...");
await delay(1000);
console.log("Done after 1 second!");
```

```javascript
// Async function example
async function fetchData() {
    await delay(500);
    return { status: "success", data: [1, 2, 3] };
}

await fetchData()
```

## Classes

```javascript
class Point {
    constructor(x, y) {
        this.x = x;
        this.y = y;
    }
    
    distance(other) {
        const dx = this.x - other.x;
        const dy = this.y - other.y;
        return Math.sqrt(dx * dx + dy * dy);
    }
    
    toString() {
        return `Point(${this.x}, ${this.y})`;
    }
}
```

```javascript
const p1 = new Point(0, 0);
const p2 = new Point(3, 4);
p1.distance(p2)
```

## Documentation

JSDoc comments document the functions. Put the cursor on a name and press **Shift+Tab**, or open **Help > Show Contextual Help**, to read its documentation. The built-in objects link to their page on MDN, like `Math.PI` below.

```javascript
/**
 * Compute the area of a circle.
 * @param {number} radius - The radius of the circle
 * @returns {number} The area of the circle
 */
function circleArea(radius) {
    return Math.PI * radius ** 2;
}

circleArea(2)
```
